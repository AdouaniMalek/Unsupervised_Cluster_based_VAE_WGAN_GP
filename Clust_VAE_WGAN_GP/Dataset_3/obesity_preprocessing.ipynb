```python
# Import necessary libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder, MinMaxScaler

# Load the dataset
dataset_directory = "C:/Users/Mega Pc/Desktop/SF-GAN unsupervised/Dataset_3/"
df = pd.read_csv(dataset_directory + "ObesityDataSet_raw_and_data_sinthetic.csv")

# Display basic information about the dataset
print("Dataset Info:")
print(df.info())
print("\nDataset Description:")
print(df.describe())

# Check for missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Visualize missing data
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
plt.title("Missing Data Heatmap")
plt.show()

# Check for duplicate rows
print("\nNumber of duplicate rows:", df.duplicated().sum())

# Drop duplicates if any
df = df.drop_duplicates()

# EDA - Visualize distributions and relationships
# Gender distribution
sns.countplot(data=df, x='Gender')
plt.title("Gender Distribution")
plt.show()

# Target variable distribution (NObeyesdad)
sns.countplot(data=df, x='NObeyesdad', order=df['NObeyesdad'].value_counts().index)
plt.title("Target Variable Distribution")
plt.xticks(rotation=45)
plt.show()

# Correlation heatmap for numerical features
numeric_features = df.select_dtypes(include=['float64', 'int64']).columns
plt.figure(figsize=(10, 8))
sns.heatmap(df[numeric_features].corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()

# Boxplot for Age grouped by Gender
sns.boxplot(data=df, x='Gender', y='Age')
plt.title("Boxplot of Age by Gender")
plt.show()

# Data preprocessing
# Encode categorical variables
categorical_columns = ['Gender', 'family_history_with_overweight', 'FAVC', 'CAEC', 'SMOKE', 
                       'SCC', 'CALC', 'MTRANS', 'NObeyesdad']
label_encoders = {}
for col in categorical_columns:
    le = LabelEncoder()
    # Use .loc to ensure proper assignment
    df.loc[:, col] = le.fit_transform(df[col])
    label_encoders[col] = le  # Save encoders for future use

# Standardize numerical features
scaler = MinMaxScaler()
numerical_columns = ['Age', 'Height', 'Weight', 'FCVC', 'NCP', 'CH2O', 'FAF', 'TUE']
df.loc[:, numerical_columns] = scaler.fit_transform(df[numerical_columns])

# Save the preprocessed data
preprocessed_path = dataset_directory + "preprocessed_obese_data.csv"
df.to_csv(preprocessed_path, index=False)
print(f"Preprocessed data saved to {preprocessed_path}")
```
